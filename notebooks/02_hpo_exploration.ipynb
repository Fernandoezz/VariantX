```python
import pandas as pd
import pronto

print("Everything works!")
```

```python
ontology = pronto.Ontology("../data/raw/hpo/hp.obo")

print(len(list(ontology.terms())))
```

```python
rows = []

for term in ontology.terms():
    rows.append({
        "HPO_ID": term.id,
        "Name": term.name,
        "Definition": term.definition
    })

hpo_df = pd.DataFrame(rows)
```

```python
hpo_df.head()
```

```python
hpo_df.to_csv(
    "../data/interim/hpo_tables/hpo_terms.csv",
    index=False
)
```
