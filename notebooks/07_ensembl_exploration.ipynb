```python
rows = []

with gzip.open(
    "../data/raw/ensembl/Homo_sapiens.GRCh38.116.gtf.gz",
    "rt"
) as f:

    for line in f:

        if line.startswith("#"):
            continue

        fields = line.strip().split("\t")

        if fields[2] != "gene":
            continue

        chromosome = fields[0]
        start_position = int(fields[3])
        end_position = int(fields[4])

        attributes = fields[8]

        gene_name = None

        for item in attributes.split(";"):

            item = item.strip()

            if item.startswith("gene_name"):
                gene_name = item.split('"')[1]

        rows.append({
            "gene_symbol": gene_name,
            "chromosome": chromosome,
            "start_position": start_position,
            "end_position": end_position
        })
```

```python
import pandas as pd

ensembl_df = pd.DataFrame(rows)

ensembl_df.head()
```
